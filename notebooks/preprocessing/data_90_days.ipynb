```python
test_df = pd.read_csv("/root/MIMICIV/data/splitted/landmark_evo_test.csv")
train_df = pd.read_csv("/root/MIMICIV/data/splitted/landmark_evo_train.csv")
vali_df = pd.read_csv("/root/MIMICIV/data/splitted/landmark_evo_vali.csv")
patients_df = pd.read_csv("/root/MIMICIV/data/mimic-iv-3.1/hosp/patients.csv")
```

```python
def dead_90_days_from_landmark(row):
    """function for determining whether a patients is died after 90 days from that specific landmark"""
    if row['death_in_90days'] == 0:
        # Here the patiend did not die
        return 0
    else:
        # Here the patient died in 90 from the very last visit
        delta_days = (pd.to_datetime(row['dod']) - pd.to_datetime(row['dischtime'])).days
        if delta_days <= 90:
            return 1
        else:
            return 0
```
